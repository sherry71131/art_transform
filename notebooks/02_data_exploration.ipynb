# 02 - Data Exploration

This notebook:
- Loads sample images from processed train/val folders.
- Visualizes a few examples from each domain (photos, Japanese art, anime).
- Confirms that preprocessing worked correctly.


```python
# import 
import os
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

PROCESSED_DIR = Path("../data/processed")
```

```python
# show a grid of images
def show_samples(folder, n=5, title="Samples"):
    folder = Path(folder)
    image_paths = list(folder.glob("*"))[:n]

    plt.figure(figsize=(12, 3))
    for i, img_path in enumerate(image_paths):
        img = Image.open(img_path)
        plt.subplot(1, n, i + 1)
        plt.imshow(img)
        plt.axis("off")
        plt.title(img_path.name[:10])
    plt.suptitle(title)
    plt.show()
```

```python
# show photos
show_samples(PROCESSED_DIR / "photos" / "train", n=5, title="Photos - Train")
```

```python
# show Japanese art / WikiArt subset
show_samples(PROCESSED_DIR / "japanese" / "train", n=5, title="Japanese Art / WikiArt - Train")
```

```python
# show anime
show_samples(PROCESSED_DIR / "anime" / "train", n=5, title="Anime / Danbooru - Train")
```
